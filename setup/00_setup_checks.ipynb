## Check Access to Cloud Storage

```sql
%fs ls 'abfss://healthcare@sthrrpdevdp.dfs.core.windows.net/'
```

## Create Catalog

```sql
%sql
CREATE CATALOG IF NOT EXISTS healthcare
  MANAGED LOCATION 'abfss://healthcare@sthrrpdevdp.dfs.core.windows.net/managed';
```

## Create Schemas and Landing Volume

```sql
%sql
CREATE SCHEMA IF NOT EXISTS healthcare.landing;
CREATE SCHEMA IF NOT EXISTS healthcare.bronze;
CREATE SCHEMA IF NOT EXISTS healthcare.silver;
CREATE SCHEMA IF NOT EXISTS healthcare.gold;
CREATE SCHEMA IF NOT EXISTS healthcare.ops;

CREATE EXTERNAL VOLUME IF NOT EXISTS healthcare.landing.raw_files
  LOCATION 'abfss://healthcare@sthrrpdevdp.dfs.core.windows.net/landing';
```

## Verify Schemas

```sql
%sql
SHOW SCHEMAS IN healthcare;
```
